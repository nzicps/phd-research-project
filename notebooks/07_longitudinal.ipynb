## Panel fixed-effects model

The descriptive summaries above show *between*-year patterns. To use the
panel structure itself -- controlling for time-invariant person-level
unobserved heterogeneity -- fit a fixed-effects regression (the primary
longitudinal estimation approach in `docs/methodology.md`).

```python
from src.panel_models import fit_fixed_effects, fixed_effects_summary_table

fe_results = fit_fixed_effects(
    df, dependent='income', exog=['chronic_condition'],
    entity_effects=True, time_effects=True,
)
fixed_effects_summary_table(fe_results)
```
